```python
#Tugas Kelompok

import pandas as pd

# 1. DATA LOADING & INSPECTION
df = pd.read_csv('dataset_penjualan_kantin.csv')
print('=== 1. INSPEKSI DATASET AWAL ===')
print('Head (5 Baris Pertama):')
print(df.head())

print('Info Dataset:')
print(df.info())

print('Describe Dataset:')
print(df.describe(include='all'))

print('Shape (Jumlah Baris & Kolom):')
print(df.shape)

# 2. DATA CLEANING
df_clean = df.copy()
# A. Hapus Duplikat
df_clean = df_clean.drop_duplicates()

# B. Penyeragaman Teks Kategori
df_clean['kategori'] = df_clean['kategori'].str.capitalize()

# C. Pembersihan Format String Rp dan Tipe Data pada harga_satuan
df_clean['harga_satuan'] = (
    df_clean['harga_satuan']
    .astype(str)
    .str.replace('Rp', '', regex=False)
    .str.replace('.', '', regex=False)
    .str.strip()
)
df_clean['harga_satuan'] = pd.to_numeric(
    df_clean['harga_satuan'], errors='coerce'
)

# D. Pembersihan Imbuhan pcs dan Tipe Data pada jumlah_terjual
df_clean['jumlah_terjual'] = (
    df_clean['jumlah_terjual']
    .astype(str)
    .str.replace('pcs', '', regex=False)
    .str.strip()
)
df_clean['jumlah_terjual'] = pd.to_numeric(
    df_clean['jumlah_terjual'], errors='coerce'
)

# E. Menangani Missing Value (Fillna)
df_clean['jumlah_terjual'] = df_clean['jumlah_terjual'].fillna(0)

# Isi harga_satuan yang NaN berdasarkan median harga produk sejenis
harga_median = df_clean.groupby('nama_produk')['harga_satuan'].median()
df_clean['harga_satuan'] = df_clean['harga_satuan'].fillna(
    df_clean['nama_produk'].map(harga_median)
)
print('=== 2. PEMBERSIHAN DATA SELESAI ===')
print(df_clean.info())


# 3. DATA MANIPULATION
# A. Kolom Turunan
df_clean['total_pendapatan'] = (
    df_clean['harga_satuan'] * df_clean['jumlah_terjual']
)

# B. Filtering (Transaksi jumlah_terjual > 5)
transaksi_besar = df_clean[df_clean['jumlah_terjual'] > 5]

# C. Sorting (Pendapatan Tertinggi)
df_sorted = df_clean.sort_values(by='total_pendapatan', ascending=False)

# D. Groupby / Agregasi (Pendapatan per Menu)
pendapatan_per_menu = (
    df_clean.groupby('nama_produk')['total_pendapatan']
    .sum()
    .reset_index()
    .sort_values(by='total_pendapatan', ascending=False)
)

# Groupby per Kategori
pendapatan_per_kategori = (
    df_clean.groupby('kategori')['total_pendapatan']
    .sum()
    .reset_index()
    .sort_values(by='total_pendapatan', ascending=False)
)

print('=== 3. HASIL MANIPULASI DATA ===')
print('Pendapatan per Menu:')
print(pendapatan_per_menu)
print('Pendapatan per Kategori:')
print(pendapatan_per_kategori)

# 4. SIMPAN DATASET BERSIH
df_clean.to_csv('dataset_bersih.csv', index=False)
print("[SUCCESS] Dataset bersih berhasil disimpan dengan nama 'dataset_bersih.csv'!")

# 5. DATA PROFILING SUMMARY
"""
RINGKASAN TEMUAN UTAMA:
1. 'Roti Bakar' dan 'Nasi Goreng' merupakan kontributor pendapatan terbesar di kantin sekolah.
2. Kategori Makanan secara keseluruhan menyumbang propo00000000000000000rsi pendapatan tertinggi dibandingkan Minuman dan Snack.
3. Setelah proses pembersihan data (menghapus format teks 'Rp'/'pcs', duplikat, dan penanganan missing value), dataset berkurang dari 69 baris mentah menjadi 65 baris data bersih yang valid.
"""
```
