Jawaban Latihan 6: Menu yang menghasilkan total_pendapatan tertinggi adalah Nasi Goreng. Informasi ini berguna untuk memastikan stok bahan baku dan marancang strategi promosi atau paket penjualan yang lebih menguntungkan.

```python
import pandas as pd
# 1. Data Loading & Inspection
df = pd.read_csv('data_kantin.csv')
print("--- 5 Baris Pertama Data ---")
print(df.head())
print("--- Informasi Struktur Data ---")
print(df.info())
print("--- Statistik Ringkas ---")
print(df.describe())
print("\n--- Dimensi Dataset (Baris, Kolom) ---")
print(df.shape)

# 2. Data Cleaning
df['terjual'] = df['terjual'].fillna(0)    # Terjual diisi dengan 0 karena diasumsikan belum ada transaksi
df = df.dropna(subset=['menu'])     # Menghapus baris tanpa nama menu karena data tidak memiliki identitas produk
df = df.drop_duplicates()     # Menangani duplikat
df['harga'] = df['harga'].astype(int)       # Memastikan tipe data sesuai
df['terjual'] = df['terjual'].astype(int)

# 3. Data Manipulation
# Kolom Turunan: Total Pendapatan
df['total_pendapatan'] = df['harga'] * df['terjual']

# Filtering: Menu Laris (Terjual > 20)
menu_laris = df[df['terjual'] > 20]

# Sorting: Pengurutan Berdasarkan Total Pendapatan Tertinggi
df_sorted = df.sort_values(by='total_pendapatan', ascending=False)

# Groupby / Agregasi: Total Pendapatan Per Menu
ringkasan_pendapatan = (
    df.groupby('menu')['total_pendapatan'].sum().reset_index()
)

# 4 Menyimpan Dataset Bersih
df.to_csv('dataset_bersih.csv', index=False)
print("\n[INFO] Dataset bersih berhasil disimpan ke 'dataset_bersih.csv'")

```
